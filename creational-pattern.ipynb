# Creational Patterns

Focuses more towards creation of object.

## Factory Method

Create objects in superclass, but allows subclasses to alter the type of the object.

```python
class Transport:
    
    def __init__(self) -> None:
        pass
    
    def deliver():
        pass
    
class Truck(Transport):
    
    def deliver():
        print('Delivered by truck!')

class Ship(Transport):
    
    def deliver():
        print('Delivered by ship!')
        
```
