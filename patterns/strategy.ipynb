```python
# Strategy Design
class Vehicle():
    
    def drive(self):
        print('Normal Strategy')

  
class Car(Vehicle):
    pass

 
class Motorcycle(Vehicle):
    
    def drive(self):
        print('Bike Strategy')
 
class Bicycle(Vehicle):
    
    def drive(self):
        print('Cycling Strategy')


class Truck(Vehicle):
    pass


class SuperCar(Vehicle):

    def drive(self):
        print('Power Strategy')


class SuperCar(Vehicle):

    def drive(self):
        # Duplicate code
        print('Power Strategy')


class OffRoadCar(Vehicle):

    def drive(self):
        # Duplicate code
        print('Power Strategy')
```

```python
import abc
class DriveStrategy:

    @abc.abstractmethod
    def drive(self):
        pass


class NormalStrategy(DriveStrategy):

    def drive(self):
        print('Normal Strategy')


class PowerStrategy(DriveStrategy):

    def drive(self):
        print('Power Strategy')


class XYZStrategy(DriveStrategy):

    def drive(self):
        print('XYZ Strategy')
```

```python
# Strategy Design
class Vehicle():
    
    __drive_class__ = None

    def fuel(self):
        print('Fueling...')

    def drive(self):
        self.__drive_class__().drive()

  
class Car(Vehicle):
    # Injecting class/obj in the parent
    __drive_class__ = NormalStrategy

 
class Motorcycle(Vehicle):
    # Injecting class/obj in the parent
    __drive_class__ = NormalStrategy


class Truck(Vehicle):
    # Injecting class/obj in the parent
    __drive_class__ = NormalStrategy


class SuperCar(Vehicle):
    # Injecting class/obj in the parent
    __drive_class__ = PowerStrategy


class OffRoadCar(Vehicle):
    # Injecting class/obj in the parent
    __drive_class__ = PowerStrategy
```
