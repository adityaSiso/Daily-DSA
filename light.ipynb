### Valid Anagram 

*Desc:* Given two strings s and t, return true if the two strings are anagrams of each other, otherwise return false.

An anagram is a string that contains the exact same characters as another string, but the order of the characters can be different.

```python
def is_anagrams_using_sorting(s: str, t: str) -> bool:
    sorted_s = ''.join(sorted(s))   # Use storting algo here instead of function.
    sorted_t = ''.join(sorted(t))
    
    return sorted_s == sorted_t

def is_anagrams_using_map(s: str, t: str) -> bool:
    
    if len(s) != len(t):
        return False
    
    count_s, count_t = {}, {}
    
    for i in s:
        count_s[i] = 1 + count_s.get(i, 0)
    
    for i in t:
        count_t[i] = 1 + count_t.get(i, 0)
            
    for i in count_s:
        if count_s[i] != count_t.get(i, 0):
            return False
        
    return True

s = "aasdasd"
t = "asdasd"
print(f"Are `{s}` and `{t}` anagrams: {is_anagrams_using_sorting(s, t)}")
print(f"Are `{s}` and `{t}` anagrams: {is_anagrams_using_map(s, t)}")
```

## Group all the Anagrams

```python
from collections import defaultdict
from typing import List
def get_anagram_groups_via_sorting(lst: List[str]) -> List[List[str]]:
    
    h_map = defaultdict(list)
    for s in lst:
        sorted_str = ''.join(sorted(s))
        h_map[sorted_str].append(s)
    
    return list(h_map.values())

def get_anagram_groups_via_hmap(lst: List[str]) -> List[List[str]]:
    
    h_map = defaultdict(list)
    for s in lst:
        count = [0] * 26            # a ... z
        for c in s:
            count[ord(c) - ord('a')] += 1  # (0, 1, 0, * 26) this will be the key of hash map.
        h_map[tuple(count)].append(s)
    
    return list(h_map.values())
strs = ["act","pots","tops","cat","stop","hat"]
print(get_anagram_groups_via_sorting(strs))
print(get_anagram_groups_via_hmap(strs))
```

## Top K frequent elements

*Desc:* Given an integer array nums and an integer k, return the k most frequent elements within the array.

The test cases are generated such that the answer is always unique.

You may return the output in any order.

```python
def k_most_frequest(lst: list, k: int) -> list: 
    
    h_map = {}
    freq = [[] for i in range(len(lst)+1)]
    for val in lst:
        h_map[val] = 1 + h_map.get(val, 0)
        
    for n, c in h_map.items():
        freq[c].append(n)

    res = []
    for i in range(len(freq) - 1, 0, -1):
        for j in freq[i]:
            res.append(j)
            if len(res) == k:
                return res

lst = [1,2,2,3,3,3]
k = 3
print(k_most_frequest(lst, k))
```

```python
def get_two_sum(nums: list, target: int) -> list:

    for i in range(0, len(nums) - 1):
        for j in range(i, len(nums)):
            if nums[i] + nums[j] == target and i != j:
                return [i, j]

def get_two_sum_using_map(nums: list, target: int) -> list:
    h_map = {}
    for idx, val in enumerate(nums):
        diff = target - val
        if diff in h_map:
            return [h_map[diff], idx]
        h_map[val] = idx
    return 
        
nums = [3,4,5,6]
target = 7
print(get_two_sum(nums, target))
print(get_two_sum_using_map(nums, target))
```

```python
def binary_search(lst: list, target: int) -> int:
    start = 0
    end = len(lst) - 1

    while start <= end:
        mid = start + (end - start)//2          # This is to prevent integer overflowing.
        
        if lst[mid] == target:
            return mid
        
        elif lst[mid] < target:
            start = mid + 1
            
        elif lst[mid] > target:
            end = mid - 1
            
    return -1

lst = [i for i in range(1110, 1123)]
target = 1122
print("Found index: ", binary_search(lst, target))
```

```python
def product_sol_1(lst: list) -> list:
    
    res = 1
    zero_count = 0
    for i in lst:
        if not i:
            zero_count += 1 
        else:
            res *= i
    
    if zero_count > 1:
        return [0] * len(nums)        
    
    for i in range(0, len(lst)):
        if lst[i]:
            if not zero_count:
                lst[i] = res // lst[i]
            else:
                lst[i] = 0
        
        else:
            lst[i] = res
            
    return lst

def product_sol_2(lst: list) -> list:
    
    n = len(lst)
    res = [0] * n
    pref = [0] * n
    post = [0] * n
    
    pref[0] = post[n - 1] = 1
    for i in range(1, n):
        pref[i] = pref[i - 1] * lst[i - 1]
    
    for i in range(n - 2, -1, -1):
        post[i] = post[i + 1] * lst[i + 1]
        
    for i in range(0, n):
        res[i] = post[i] * pref[i]
        
    return res
        
lst = [1, 2, 0, 6, 19]
product_sol_1(lst[:]), product_sol_2(lst[:])
```

## Longest Consecutive Sequence
Given an array of integers nums, return the length of the longest consecutive sequence of elements that can be formed.

A consecutive sequence is a sequence of elements in which each element is exactly 1 greater than the previous element. The elements do not have to be consecutive in the original array.

You must write an algorithm that runs in O(n) time.

```python
def cons_seq_sol_1(nums: list) -> int:
    n = len(nums)
    h_set = set(nums)
    max_streak = -1
    
    for n in nums:
        streak, num = 0, n
        while num in h_set:
            num += 1
            streak += 1
        max_streak = max(streak, max_streak)    
        
    print(max_streak)
    
lst = [2, 20, 4, 10, 3, 4, 5, 9, 6]
cons_seq_sol_1(lst)
```

# 3Sum
Given an integer array nums, return all the triplets [nums[i], nums[j], nums[k]] where nums[i] + nums[j] + nums[k] == 0, and the indices i, j and k are all distinct.

The output should not contain any duplicate triplets. You may return the output and the triplets in any order.

```python
def three_sum_sol_1(nums: list) -> list:
    
    n = len(nums)
    res = set()
    
    for i in range(n-1):
        h_set = set()
        for j in range(i+1, n):
            third = -(nums[i] + nums[j])
            if third in h_set:
                res.add(tuple(sorted([nums[i], nums[j], third])))
            h_set.add(nums[j])
    return res

def three_sum_sol_2(nums: list) -> list:
    
    n = len(nums)
    storted_nums = sorted(nums)
    res = []
    i = 0
    
    for i in range(n):
        if i > 0 and storted_nums[i] == storted_nums[i-1]:
            continue
        j = i + 1
        k = n - 1
        while j < k:
            sum_ = storted_nums[i] + storted_nums[j] + storted_nums[k]
            if not sum_:
                res.append([storted_nums[i], storted_nums[j], storted_nums[k]])
                j += 1
                k -= 1

                while (j < k and storted_nums[j] == storted_nums[j - 1]): j += 1
                while (k > j and storted_nums[k] == storted_nums[k + 1]): k -= 1
            
            elif sum_ < 0:
                j += 1
            
            elif sum_ > 0:
                k -= 1
    return res

nums = [-1,0,1,2,-1,-4] 
three_sum_sol_1(nums), three_sum_sol_2(nums)
```

```python
# Stock Span Problem

def brute(lst: list) -> list:

    n: int = len(lst)
    res: list = []

    for i in range(n):
        count: int = 0
        for j in range(i - 1, -1, -1):
            if lst[i] > lst[j]:
                count += 1
            
            else:
                break
        res.append(count + 1)
    
    return res

def optimized(lst: list) -> list:

    n: int = len(lst)
    res: list = []
    stack: list = []

    for i in range(n):

        if not stack:
            res.append(1)

        elif lst[i] < stack[-1][0]:
            res.append(i - stack[-1][1]) 

        elif stack[-1][0] < lst[i]:

            while stack and lst[i] > stack[-1][0]:
                stack.pop()

            if not stack:
                res.append(1)

            elif lst[i] < stack[-1][0]:
                res.append(i - stack[-1][1])

        stack.append((lst[i], i))
    
    return res
            
stocks = [100, 80, 60, 70, 60, 75, 85]
brute(stocks), optimized(stocks)
```

```python
# Rain Water Trapping

def get_left_max(buildings: list) -> list:

    n = len(buildings)
    left = [buildings[0]]
    for i in range(1, n):
        left.append(max(left[i-1], buildings[i]))
    return left

def get_right_max(buildings: list) -> list:

    n = len(buildings)
    right = [buildings[-1]]
    for i in range(n - 2, -1, -1):
        # Note this... right list element starts from 0 but loop is backwards.
        right.append(max(right[(n - 1 - i - 1)], buildings[i]))
    return right[::-1]

def brute(lst: list) -> int:
    
    n = len(lst)
    area = 0
    for i in range(1, n - 1):
        left_max = max(lst[:i])
        right_max = max(lst[i+1:])

        if (min(left_max, right_max) - lst[i]) > 0:
            area = area + min(left_max, right_max) - lst[i]

    return area

def optimized(buildings: list) -> int:

    n = len(buildings)
    left_max = get_left_max(buildings)
    right_max = get_right_max(buildings)
    sum_ = 0

    for i in range(n):
        sum_ = sum_ + min(left_max[i], right_max[i]) - buildings[i]

    return sum_
buildings = [0, 1, 0, 2, 1, 0, 1, 3, 2, 1, 2, 1]
brute(buildings), optimized(buildings)
```

```python
# Min element in a stack O(n) space

class MinElement:

    stack1 = []
    min_stack = []

    def _push(self, element):
        self.stack1.append(element)
        if not self.min_stack or self.min_stack[-1] >= element:
            self.min_stack.append(element)

    def _pop(self):
        if not self.stack1:
            return -1
        
        else:
            poped = self.stack1.pop()
            if poped == self.min_stack[-1]:
                self.min_stack.pop()

    def get_min(self):

        if not self.min_stack:
            return -1
        return self.min_stack[-1]
    
obj = MinElement()
```

```python
# Find Kth minimum element
from queue import PriorityQueue

def heap(lst: list, k: int) -> int:

    n = len(lst)
    # Creates a queue with lower on top always.
    # IN MAX HEAP MIN IN ON THE BOTTOM.
    max_heap = PriorityQueue()

    for i in range(n):
        
        # Multiply with -1 to reverse the heap order.
        max_heap.put(lst[i] * -1)
        
        if max_heap.qsize() > k:
            max_heap.get()
        
    return abs(max_heap.get())

lst = [7, 10, 4, 3, 15, 20]
heap(lst, 3)
```

```python
# Sort K sorted array.
from queue import PriorityQueue

def heap(lst: list, k: int) -> int:

    n = len(lst)
    # Creates a queue with lower on top always.
    min_heap = PriorityQueue()
    res = []

    for i in range(n):

        if min_heap.qsize() > k:
            res.append(min_heap.get())
        min_heap.put(lst[i])

    while min_heap.qsize() > 0:
        res.append(min_heap.get())

    return res

lst = [6, 5, 3, 2, 8, 10, 9]
heap(lst, 3)
```

```python
# K Closest Point to Origin

from queue import PriorityQueue
import math

def optimized(points, k):

    distance = {}
    max_queue = PriorityQueue()

    for i, j in points:
       distance[(i, j)] = (math.sqrt(i**2 + j**2)) * -1

    for point, dist in distance.items():
        max_queue.put((dist, point))

        if max_queue.qsize() > k:
            max_queue.get()
        
    return [point for _, point in max_queue.queue]

points = [(0, 1), (4, 0), (3, -1), (2, 1), (2, 2)]
k = 2
optimized(points, k)
```

```python
# Connect ropes to minimise the cost

from queue import PriorityQueue

def optimized(ropes):

    cost = 0
    min_queue = PriorityQueue()
    for rope in ropes:
        min_queue.put(rope)

    while min_queue.qsize() >= 2:

        r1 = min_queue.get()
        r2 = min_queue.get()
        r3 = r2 + r1
        min_queue.put(r3)
        cost += r3

    return cost

ropes = [1, 2, 3, 4, 5]
optimized(ropes)
```

<h1> Linked List </h1>

```python
# You always starts with the head.
class Node:

    def __init__(self, value, head = None):
        self.value = value
        self.next = None


class LinkedList:

    def __init__(self):
        self.head = None

    def insert_at_end(self, data):

        new_node = Node(data)

        if not self.head:
            self.head = new_node
            return

        current = self.head
        while current.next:
            current = current.next
        current.next = new_node

    def insert_at_beginning(self, data):

        new_node = Node(data)
        new_node.next = self.head
        self.head = new_node

    def delete_node(self, key):
        
        current = self.head

        if current and current.value == key:
            self.head = current.next
            del current
            return
        
        prev = None
        while current and current.value != key:
            prev = current
            current = current.next
        
        if not current:
            print(f'No Node found with {key}...')
            return
        prev.next = current.next
        del current

    def traverse(self):

        current = self.head
        while current:
            print(f'{current.value} is pointing to {current.next}')
            current = current.next

    def lenght(self):
        
        count = 0
        current = self.head

        while current:
            count += 1
            current = current.next
        return count
```

```python
l = LinkedList()
l.insert_at_end(10)
l.insert_at_end(20)
l.insert_at_end(30)
print(f'Length of linked list is: {l.lenght()}')
l.traverse()
l.delete_node(20)
l.insert_at_beginning(5)
l.insert_at_end(40)
print(f'Length of linked list is: {l.lenght()}')
l.traverse()
```
