```python
from abc import ABCMeta, abstractmethod
class AbstractBike(metaclass=ABCMeta):
    __PRODUCT__ = 'bike'

    @abstractmethod
    def build(self):
        pass

    @abstractmethod
    def show(self):
        pass

class ConcreteBikeA(AbstractBike):
    __PRODUCT__ = 'crusier'
    
    def build(self):
        return self

    def show(self):
        return self
    
class ConcreteBikeB(AbstractBike):
    __PRODUCT__ = 'super'
    
    def build(self):
        return self

    def show(self):
        return self
    
class ConcreteBikeC(AbstractBike):
    __PRODUCT__ = 'naked'

    def build(self):
        return self

    def show(self):
        return self
```

```python
from abc import ABCMeta, abstractmethod
class AbstractCar(metaclass=ABCMeta):
    __PRODUCT__ = 'car'
    
    @abstractmethod
    def build(self):
        pass

    @abstractmethod
    def show(self):
        pass

class ConcreteCarA(AbstractCar):
    __PRODUCT__ = 'suv'
    
    def build(self):
        return self

    def show(self):
        return self
    
class ConcreteBikeB(AbstractCar):
    __PRODUCT__ = 'super'
    
    def build(self):
        return self

    def show(self):
        return self
    
class ConcreteCarC(AbstractCar):
    __PRODUCT__ = 'sedan'

    def build(self):
        return self

    def show(self):
        return self
```

```python
class AbstractFactor(metaclass=ABCMeta):
    __FACTORY__ = None

    @abstractmethod
    def show(self, type):
        pass

    @abstractmethod
    def build(self, type):
        pass

class FactorBike(AbstractFactor):
    __FACTORY__ = AbstractBike.__PRODUCT__

    def show(self, type):
        for bike in AbstractBike.__subclasses__():
            if bike.__PRODUCT__.casefold() == type.casefold():
                return bike().show()

    def build(self, type):
        for bike in AbstractBike.__subclasses__():
            if bike.__PRODUCT__.casefold() == type.casefold():
                return bike().build()

class FactorCar(AbstractFactor):
    __FACTORY__ = AbstractCar.__PRODUCT__

    def show(self, type):
        for car in AbstractCar.__subclasses__():
            if car.__PRODUCT__.casefold() == type.casefold():
                return car().show()

    def build(self, type):
        for car in AbstractCar.__subclasses__():
            if car.__PRODUCT__.casefold() == type.casefold():
                return car().build()
```

```python
# class AbstracCreator can be implemented here.
class Creator:

    @staticmethod
    def _get_product(product):
        for prod_cls in AbstractFactor.__subclasses__():
            if prod_cls.__FACTORY__.casefold() == product.casefold():
                return prod_cls()
        raise ValueError('Product not available!')

    @staticmethod
    def show(product, type):
        return Creator._get_product(product).show(type)

    @staticmethod
    def build(product, type):
        return Creator._get_product(product).build(type)
```

```python
Creator.show('car', 'sedan')
```
