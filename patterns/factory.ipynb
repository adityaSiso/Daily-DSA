```python
# Factory Pattern
import abc

class Shape:

    def __init__(self, *args, **kwargs):
        pass    
    
    @abc.abstractmethod
    def draw():
        pass

    @abc.abstractmethod
    def area():
        pass
```

```python
class Rectangle(Shape):

    def __init__(self, l, w, *args, **kwargs):
        super().__init__(*args, **kwargs)
        self.l = l
        self.w = w

    def draw(self):
        print('-'*10)
        print('|', ' '*8, '|')
        print('|', ' '*8, '|')
        print('-'*10)

    def area(self):
        print(f'Area of {self.l}X{self.w} rectangle is `{self.l * self.w}`')
```

```python
import math
class Circle(Shape):

    def __init__(self, r, *args, **kwargs):
        super().__init__(*args, **kwargs)
        self.r = r

    def draw(self):
        print('O'*10)

    def area(self):
        print(f'Area of circle with radis {self.r} is `{2 * math.pi * self.r}`')
```

```python
class Triangle(Shape):

    def __init__(self, a, b, c, *args, **kwargs):
        super().__init__(*args, **kwargs)
        self.a = a
        self.b = b
        self.c = c

    def draw(self):
        print('    /\\')
        print('   /  \\')
        print('  /    \\')
        print(' /      \\')
        print('/________\\')

    def area(self):
        print(f'Area of triangle {self.a}X{self.b}X{self.c} is `{self.a * self.b * self.c}`')
```

```python
def get_shape(input):

    for i in Shape.__subclasses__():
        if i.__name__.casefold() == input.casefold():
            return i
```

```python
obj = get_shape('rectangle')(10, 20)
```
