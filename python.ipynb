```python
class ValidateString:

    def __init__(self, min_length=0):
        print('Init called')
        self.min_l = min_length

    def __set_name__(self, owner_class, property_name):
        print(f'__set_name__ called, self={self}, owner_class={owner_class}, property={property_name}.')
        self.property_name = property_name

    def __get__(self, instance, owner_class):
        if not instance:
            return self
        print(f'__get__ called for property {self.property_name} of instance {instance}.')
        return instance.__dict__.get(self.property_name)
        # Will avoid descriptors.
        # return getattr(instance, f'_{self.property_name}')

    def __set__(self, instance, value):
        print(f'__set__ called with value={value}')
        if  not isinstance(value, str):
            raise ValueError(f'{self.property_name} should be a string.')
        if len(value) < self.min_l:
            raise ValueError(f'{self.property_name} must be at lreast {self.min_l} chars.')
        instance.__dict__[self.property_name] = value
        # Will avoid descriptors.
        # setattr(instance, f'_{self.property_name}', value)

class Person:
    f_name = ValidateString(1)
    l_name = ValidateString(2)

p = Person()
```

```python
from types import MethodType
# Decorator Class
class DecoClass:

    def __init__(self, fn):
        print(f'Decorating Function: {fn.__name__}')
        self.fn = fn

    def __call__(self, *args, **kwds):
        print(f'Calling Function with {self}, {args}, {kwds}')
        return self.fn(*args, **kwds)
    
    def __get__(self, instance, owner):
        print(f'__get__ called, self={self}, instance={instance}, owner={owner}.')
        if instance is None:
            return self
        else:
            x = MethodType(self, instance)
            return x
        
class MyClass:

    # Functions gets tranformed into a method when called from an instance becuase they implement discriptors protocol.
    @DecoClass
    def myFunc(self, a, b):
        return a + b

    def myNormalFunc(self):
        pass

my_obj = MyClass()
```

```python
def process_mesasges(msg_list: list) -> list:

    n = len(msg_list)
    parsed_msgs = []
    final = {}
    for msg in msg_list[n - 10:]:
        t, s, c = msg.split(':')
        parsed_msgs.append((int(t), int(s), c))
    t_sorted_msgs = sorted(parsed_msgs, key= lambda tup: tup[0])

    print(t_sorted_msgs)
    for msg in t_sorted_msgs:
        if (msg[0], msg[1]) not in final:
            final[(msg[0], msg[1])] = msg[2]

    return list(final.values())
```

```python
# This is a clouser, variables of outer function are `readonly` to the inner function.function
# If we try to modify outer variable in inner function without `nonlocal` keywork it will create a new local variable instead of updating the outer one.   
def outer():
    x = 'outer'
    def inner():
        # nonlocal x
        x = 'inner'
        print(x) # This makes the nested function a clouser.
    inner()
    print(x)
outer()
```

```python
import threading

# Shared variable and a lock for synchronization
current_number = 0
lock = threading.Lock()

def print_numbers(thread_id, total_threads):
    global current_number
    while current_number < 10:
        with lock:
            if current_number % total_threads == thread_id:
                print(f"Thread {thread_id}: {current_number}")
                current_number += 1

# Create 3 threads
threads = []
total_threads = 3
for i in range(total_threads):
    thread = threading.Thread(target=print_numbers, args=(i, total_threads))
    threads.append(thread)
    thread.start()

# Wait for all threads to complete
for thread in threads:
    thread.join()
```

```python
import time
from threading import Thread

def square(numbers):
    for n in numbers:
        time.sleep(0.2)
        print(f'Square of {n}: {n*n}')

def cube(numbers):
    for n in numbers:
        time.sleep(0.2)
        print(f'Cube of {n}: {n*n*n}')

numbers = [2, 3, 4, 5]
t = time.perf_counter()

# square(numbers)
# cube(numbers)

# Threading
t1 = Thread(target=square, args=(numbers,))
t2 = Thread(target=cube, args=(numbers,))

t1.start()
t2.start()

t1.join()
t2.join()

print(f'Total time taken: {time.perf_counter() - t}')
```

```python
result = []
def square(numbers):
    global result
    for n in numbers:
        result.append(n*n)

from multiprocessing import Process
p = Process(target=square, args=([1, 2, 3, 4, 5],))
p.start()
p.join()
print(f'Result: {result}')
```
