### Contains Duplicate

*Desc:* Given an integer array nums, return true if any value appears more than once in the array, otherwise return false.

```python
nums = [1, 2, 3, 8]

def does_contain_duplicate(nums: list) -> bool:
    h_map: dict = {}
    for i in nums:
        if i in h_map:
            return True
        else:
            h_map[i] = 0
    return False
        
print("Does the array contains duplicate value: ",does_contain_duplicate(nums))
```

### Valid Anagram 

*Desc:* Given two strings s and t, return true if the two strings are anagrams of each other, otherwise return false.

An anagram is a string that contains the exact same characters as another string, but the order of the characters can be different.

```python
def is_anagrams_using_sorting(s: str, t: str) -> bool:
    sorted_s = ''.join(sorted(s))   # Use storting algo here instead of function.
    sorted_t = ''.join(sorted(t))
    
    return sorted_s == sorted_t

def is_anagrams_using_map(s: str, t: str) -> bool:
    
    if len(s) != len(t):
        return False
    
    count_s, count_t = {}, {}
    
    for i in s:
        count_s[i] = 1 + count_s.get(i, 0)
    
    for i in t:
        count_t[i] = 1 + count_t.get(i, 0)
            
    for i in count_s:
        if count_s[i] != count_t.get(i, 0):
            return False
        
    return True

s = "aasdasd"
t = "asdasd"
print(f"Are `{s}` and `{t}` anagrams: {is_anagrams_using_sorting(s, t)}")
print(f"Are `{s}` and `{t}` anagrams: {is_anagrams_using_map(s, t)}")
```

## Group all the Anagrams

```python
from collections import defaultdict
from typing import List
def get_anagram_groups_via_sorting(lst: List[str]) -> List[List[str]]:
    
    h_map = defaultdict(list)
    for s in lst:
        sorted_str = ''.join(sorted(s))
        h_map[sorted_str].append(s)
    
    return list(h_map.values())

def get_anagram_groups_via_hmap(lst: List[str]) -> List[List[str]]:
    
    h_map = defaultdict(list)
    for s in lst:
        count = [0] * 26            # a ... z
        for c in s:
            count[ord(c) - ord('a')] += 1  # (0, 1, 0, * 26) this will be the key of hash map.
        h_map[tuple(count)].append(s)
    
    return list(h_map.values())
strs = ["act","pots","tops","cat","stop","hat"]
print(get_anagram_groups_via_sorting(strs))
print(get_anagram_groups_via_hmap(strs))
```

## Top K frequent elements

*Desc:* Given an integer array nums and an integer k, return the k most frequent elements within the array.

The test cases are generated such that the answer is always unique.

You may return the output in any order.

```python
def k_most_frequest(lst: list, k: int) -> list: 
    
    h_map = {}
    freq = [[] for i in range(len(lst)+1)]
    for val in lst:
        h_map[val] = 1 + h_map.get(val, 0)
        
    for n, c in h_map.items():
        freq[c].append(n)

    res = []
    for i in range(len(freq) - 1, 0, -1):
        for j in freq[i]:
            res.append(j)
            if len(res) == k:
                return res

lst = [1,2,2,3,3,3]
k = 3
print(k_most_frequest(lst, k))
```

### Two Sum
*Desc:* Given an array of integers nums and an integer target, return the indices i and j such that nums[i] + nums[j] == target and i != j.

**You may assume that every input has exactly one pair of indices i and j that satisfy the condition.**

Return the answer with the smaller index first.

```python
def get_two_sum(nums: list, target: int) -> list:

    for i in range(0, len(nums) - 1):
        for j in range(i, len(nums)):
            if nums[i] + nums[j] == target and i != j:
                return [i, j]

def get_two_sum_using_map(nums: list, target: int) -> list:
    h_map = {}
    for idx, val in enumerate(nums):
        diff = target - val
        if diff in h_map:
            return [h_map[diff], idx]
        h_map[val] = idx
    return 
        
nums = [3,4,5,6]
target = 7
print(get_two_sum(nums, target))
print(get_two_sum_using_map(nums, target))
```

## Binary Search

```python
def binary_search(lst: list, target: int) -> int:
    start = 0
    end = len(lst) - 1

    while start <= end:
        mid = start + (end - start)//2          # This is to prevent integer overflowing.
        
        if lst[mid] == target:
            return mid
        
        elif lst[mid] < target:
            start = mid + 1
            
        elif lst[mid] > target:
            end = mid - 1
            
    return -1

lst = [i for i in range(1110, 1123)]
target = 1122
print("Found index: ", binary_search(lst, target))
```

```python
def binary_search(lst: list, target: int, occ: str) -> int:
    start = 0
    end = len(lst) - 1
    result = -1

    while start <= end:
        mid = start + (end - start)//2          # This is to prevent integer overflowing.
        
        if lst[mid] == target:
            result = mid
            if occ == 'first':
                end = mid - 1
            else:
                start = mid + 1
        
        elif lst[mid] < target:
            start = mid + 1                
            
        elif lst[mid] > target:
            end = mid - 1
            
    return result

lst = [5, 7, 9, 12, 12, 12, 12, 17, 20]
target = 12
print("Found index: ", binary_search(lst, target, 'first'))
```

## Products of Array Except Self
Given an integer array nums, return an array output where output[i] is the product of all the elements of nums except nums[i].

Each product is guaranteed to fit in a 32-bit integer.

Follow-up: Could you solve it in 
O

O(n) time without using the division operation?

```python
def product_sol_1(lst: list) -> list:
    
    res = 1
    zero_count = 0
    for i in lst:
        if not i:
            zero_count += 1 
        else:
            res *= i
    
    if zero_count > 1:
        return [0] * len(nums)        
    
    for i in range(0, len(lst)):
        if lst[i]:
            if not zero_count:
                lst[i] = res // lst[i]
            else:
                lst[i] = 0
        
        else:
            lst[i] = res
            
    return lst

def product_sol_2(lst: list) -> list:
    
    n = len(lst)
    res = [0] * n
    pref = [0] * n
    post = [0] * n
    
    pref[0] = post[n - 1] = 1
    for i in range(1, n):
        pref[i] = pref[i - 1] * lst[i - 1]
    
    for i in range(n - 2, -1, -1):
        post[i] = post[i + 1] * lst[i + 1]
        
    for i in range(0, n):
        res[i] = post[i] * pref[i]
        
    return res
        
lst = [1, 2, 0, 6, 19]
product_sol_1(lst[:]), product_sol_2(lst[:])
```

## Longest Consecutive Sequence
Given an array of integers nums, return the length of the longest consecutive sequence of elements that can be formed.

A consecutive sequence is a sequence of elements in which each element is exactly 1 greater than the previous element. The elements do not have to be consecutive in the original array.

You must write an algorithm that runs in O(n) time.

```python
def cons_seq_sol_1(nums: list) -> int:
    n = len(nums)
    h_set = set(nums)
    max_streak = -1
    
    for n in nums:
        streak, num = 0, n
        while num in h_set:
            num += 1
            streak += 1
        max_streak = max(streak, max_streak)    
        
    print(max_streak)
    
lst = [2, 20, 4, 10, 3, 4, 5, 9, 6]
cons_seq_sol_1(lst)
```

# 3Sum
Given an integer array nums, return all the triplets [nums[i], nums[j], nums[k]] where nums[i] + nums[j] + nums[k] == 0, and the indices i, j and k are all distinct.

The output should not contain any duplicate triplets. You may return the output and the triplets in any order.

```python
def three_sum_sol_1(nums: list) -> list:
    
    n = len(nums)
    res = set()
    
    for i in range(n-1):
        h_set = set()
        for j in range(i+1, n):
            third = -(nums[i] + nums[j])
            if third in h_set:
                res.add(tuple(sorted([nums[i], nums[j], third])))
            h_set.add(nums[j])
    return res

def three_sum_sol_2(nums: list) -> list:
    
    n = len(nums)
    storted_nums = sorted(nums)
    res = []
    i = 0
    
    for i in range(n):
        if i > 0 and storted_nums[i] == storted_nums[i-1]:
            continue
        j = i + 1
        k = n - 1
        while j < k:
            sum_ = storted_nums[i] + storted_nums[j] + storted_nums[k]
            if not sum_:
                res.append([storted_nums[i], storted_nums[j], storted_nums[k]])
                j += 1
                k -= 1

                while (j < k and storted_nums[j] == storted_nums[j - 1]): j += 1
                while (k > j and storted_nums[k] == storted_nums[k + 1]): k -= 1
            
            elif sum_ < 0:
                j += 1
            
            elif sum_ > 0:
                k -= 1
    return res

nums = [-1,0,1,2,-1,-4] 
three_sum_sol_1(nums), three_sum_sol_2(nums)
```

## Container With Most Water
You are given an integer array heights where heights[i] represents the height of the

You may choose any two bars to form a container. Return the maximum amount of water a container can store.

```python
def water_container_sol_1(nums: list) -> int:
    
    n = len(nums)
    max_ = 0
    
    for i in range(n):
        for j in range(i+1, n):
            height = min(nums[j], nums[i])
            length = j - i
            # print(f'Height={height}, Length={length}')
            max_ = max(max_, height * length)
    return max_

def water_container_sol_2(heights: list) -> int:
    
    n = len(heights)
    i, j = 0, n - 1
    res = 0
    
    while i < j:
        
        height = min(heights[i], heights[j])
        lenght = j - i
        res = max(res, height * lenght)
        
        if heights[i] <= heights[j]:
            i += 1
        
        else: 
            j -= 1
    
    return res

heights = [1,7,2,5,4,7,3,6]
water_container_sol_1(heights), water_container_sol_2(heights)
```
