```python
class Person:
    
    def __init__(self, name: str) -> None:
        print(f'Hello {name} from {self}!')
        self.name = name
        
    def __new__(cls, name):
        instance = super().__new__(cls)
        print(f'Creating object {instance}.')
        return instance
        
p = Person('Max')
```

```python
class Person:
    
    def __init__(self, name: str) -> None:
        print(f'Hello {name} from {self}!')
        self.name = name
        
    def __new__(cls, name):
        instance = super().__new__(cls)
        print(f'Creating object {instance}.')
        return instance
    
    def register(self, func):
        setattr(self, func.__name__, MethodType(func, self))
        
p = Person('Max')
```

```python
def myFunc(self):
    print(f'This is a registered method by {self.name}.')
    
p.register(myFunc)
```

```python
p.__dict__
```

```python
p.myFunc()
```

```python
stella = Person('Stella')
stella.register(myFunc)
```

```python
stella.myFunc()
```

```python
class Person:
    
    def __init__(self, name: str, salary: int) -> None:
        print(f'Hello {name} from {self}!')
        self.name = name
        self._salary = salary
    
    @property
    def salary(self):
        return f'{self.name} salary is {"*"*len(str(self._salary))}.'
    
    @salary.setter
    def salary(self, value):
        if not value or value < 0:
            raise ValueError('Boooomm....')
        self._salary = value
         
    def __new__(cls, name, salary):
        instance = super().__new__(cls)
        print(f'Creating object {instance}.')
        return instance
        
p = Person('Max', 10000)
```

```python
p.__dict__
```

```python
Person.__dict__
```

```python
p.salary
```

```python
stella = Person('Stella', 1000000)
stella.salary
```

```python
class MyClass:
    
    def say_hello():
        print('Hello World!')
    
    @classmethod
    def say_gm(self):
        print('Good morning!')
    
    @staticmethod
    def say_bye(self):
        print('Bye Bye!')
        
obj = MyClass()
```

```python
obj.__dict__, MyClass.__dict__
```

```python
MyClass.say_gm, MyClass.say_hello, MyClass.say_bye
```

```python
obj.say_bye, obj.say_gm, obj.say_hello
```

```python
class Person:
    
    def __init__(self, name: str, salary: int) -> None:
        print(f'Hello {name}!!!')
        self.name = name
        self._salary = salary
        
    def __repr__(self) -> str:
        return f'{self.__class__.__name__}(name={self.name}, salary={self._salary})'
    
    def __str__(self) -> str:
        return f'Yooo... what is uppp {self.name}!!!'
    
    @property
    def salary(self):
        return f'{self.name} salary is {"*"*len(str(self._salary))}.'
    
    @salary.setter
    def salary(self, value):
        if not value or value < 0:
            raise ValueError('Boooomm....')
        self._salary = value
         
    def __new__(cls, name, salary):
        instance = super().__new__(cls)
        print(f'Creating object...')
        return instance
        
p = Person('Max', 10000)
```
